# Estadisticas defensivas

```python
import matplotlib.pyplot as plt
import pandas as pd
import ipywidgets as widgets
from IPython.display import display


# Load data
df_defensive = pd.read_csv("C:/UABC/Topicos selectos de la investigacion/MatchesDataSet/data_sets/estadisticas_defensivas_2017_2024.txt")
df_defensive['Temporada'] = df_defensive['Temporada'].astype(str)

# Define available seasons
seasons = sorted(df_defensive['Temporada'].unique())

# Define the metrics for tackling, renaming for clarity
metrics = {
    'Derribos Tkl': 'Derribos Tkl', 
    'Derribos 3.º def.': 'Derribos en defensa', 
    'Derribos 3.º cent.': 'Derribos en el centro del campo', 
    'Derribos 3.º ataq.': 'Derribos en ataque',
}

# Define the plotting function for a selected season
def plot_season(season):
    season_data = df_defensive[df_defensive['Temporada'] == season]
    
    # Plot 1: Tackling Metrics by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data.rename(columns=metrics).set_index('Equipo')[list(metrics.values())].plot(
        kind='bar', stacked=True, ax=ax, color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'])
    ax.set_title(f'Métricas de Derribos por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Conteo de Derribos')
    plt.xticks(rotation=45, ha='right')  # Adjust rotation and alignment
    plt.tight_layout()
    plt.show()

    # Plot 2: Interceptions and Blocks by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    ax.plot(season_data['Equipo'], season_data['Intercepciones'], label='Intercepciones', marker='o')
    ax.plot(season_data['Equipo'], season_data['Bloqueos'], label='Bloqueos', marker='o')
    ax.set_title(f'Intercepciones y Bloqueos por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Conteo')
    ax.legend()
    plt.xticks(rotation=45, ha='right')  # Adjust rotation and alignment
    plt.tight_layout()
    plt.show()

    # Plot 3: Successful Challenges and Success Rate by Team
    fig, ax1 = plt.subplots(figsize=(14, 8))
    ax2 = ax1.twinx()
    season_data.set_index('Equipo')['Desafíos Tkl'].plot(kind='bar', ax=ax1, color='skyblue', position=1, width=0.4, label='Desafíos')
    season_data.set_index('Equipo')['Desafíos Tkl%'].plot(kind='line', ax=ax2, color='darkred', marker='o', label='% Éxito')
    ax1.set_title(f'Desafíos y Porcentaje de Éxito por Equipo ({season})')
    ax1.set_xlabel('Equipo')
    ax1.set_ylabel('Número de Desafíos')
    ax2.set_ylabel('% de Éxito')
    ax1.legend(loc='upper left')
    ax2.legend(loc='upper right')
    plt.xticks(rotation=45, ha='right')  # Adjust rotation and alignment
    plt.tight_layout()
    plt.show()

    # Plot 4: Clearances and Defensive Errors by Team with Renamed Labels
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data_renamed = season_data.rename(columns={'Desp.': 'Despejes', 'Err': 'Errores'})
    season_data_renamed.set_index('Equipo')[['Despejes', 'Errores']].plot(kind='bar', color=['steelblue', 'coral'], ax=ax)
    ax.set_title(f'Despejes y Errores Defensivos por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Conteo')
    plt.xticks(rotation=45, ha='right')  # Adjust rotation and alignment
    plt.tight_layout()
    plt.show()

# Create an interactive slider widget for selecting the season
season_slider = widgets.SelectionSlider(
    options=seasons,
    description='Temporada:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True
)

# Display the slider and link it to the plotting function
interactive_plot = widgets.interactive(plot_season, season=season_slider)
display(interactive_plot)
```

# Estadisticas de creacion de oportunidades y goles

```python
import matplotlib.pyplot as plt
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# Ensure inline display of plots in Jupyter Notebook
%matplotlib inline

# Load data
df_goal_actions = pd.read_csv("C:/UABC/Topicos selectos de la investigacion/MatchesDataSet/data_sets/estadisticas_gca_2017_2024.txt")
df_goal_actions['Temporada'] = df_goal_actions['Temporada'].astype(str)

# Define unique seasons for the slider
seasons = sorted(df_goal_actions['Temporada'].unique())

# Define the plotting function for a selected season
def plot_season(season):
    season_data = df_goal_actions[df_goal_actions['Temporada'] == season]
    
    # Plot 1: Shot-Creating Actions (ACT) and Goal-Creating Actions (ACG) by Team
    fig, ax1 = plt.subplots(figsize=(14, 8))
    ax1.bar(season_data['Equipo'], season_data['ACT'], color='skyblue', label='ACT')
    ax1.bar(season_data['Equipo'], season_data['ACG'], color='salmon', bottom=season_data['ACT'], label='ACG')
    ax1.set_title(f'Acciones para la Creación de Tiros (ACT) y Goles (ACG) por Equipo ({season})')
    ax1.set_xlabel('Equipo')
    ax1.set_ylabel('Número de Acciones')
    ax1.legend()
    plt.xticks(rotation=45, ha='right')  # Adjust rotation and alignment
    plt.tight_layout()
    plt.show()

    # Plot 2: Detailed Shot-Creating Actions by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data.set_index('Equipo')[['PassLive_SCA', 'PassDead_SCA', 'HASTA_SCA', 'Dis_SCA', 'FR_SCA', 'Def_SCA']].plot(
        kind='bar', stacked=True, ax=ax, color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b'])
    ax.set_title(f'Detalle de Acciones para la Creación de Tiros (SCA) por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Número de Acciones de Creación de Tiros')
    plt.xticks(rotation=45, ha='right')  # Adjust rotation and alignment
    plt.tight_layout()
    plt.show()

    # Plot 3: Detailed Goal-Creating Actions by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data.set_index('Equipo')[['PassLive_GCA', 'PassDead_GCA', 'HASTA_GCA', 'Dis_GCA', 'FR_GCA', 'Def_GCA']].plot(
        kind='bar', stacked=True, ax=ax, color=['#17becf', '#e377c2', '#7f7f7f', '#bcbd22', '#aec7e8', '#ffbb78'])
    ax.set_title(f'Detalle de Acciones para la Creación de Goles (GCA) por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Número de Acciones de Creación de Goles')
    plt.xticks(rotation=45, ha='right')  # Adjust rotation and alignment
    plt.tight_layout()
    plt.show()

# Create an interactive slider widget for selecting the season
season_slider = widgets.SelectionSlider(
    options=seasons,
    description='Temporada:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True
)

# Link the slider to the plotting function
interactive_plot = widgets.interactive(plot_season, season=season_slider)
display(interactive_plot)
```

# Estadisticas de pases

```python
import matplotlib.pyplot as plt
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# Ensure plots display inline in Jupyter
%matplotlib inline

# Load data
df_passing = pd.read_csv("C:/UABC/Topicos selectos de la investigacion/MatchesDataSet/data_sets/estadisticas_pases_2017_2024.txt")
df_passing['Temporada'] = df_passing['Temporada'].astype(str)

# Standardize team names
df_passing['Equipo'] = df_passing['Equipo'].replace({
    'Deportivo de La Coruna': 'Deportivo La Coruna', 
    # Add other mappings if there are similar issues with other teams
})

# Sort data by team and season to maintain consistency
df_passing = df_passing.sort_values(by=['Equipo', 'Temporada'])

# Define available seasons
seasons = sorted(df_passing['Temporada'].unique())

# Define the plotting function for a selected season
def plot_season(season):
    season_data = df_passing[df_passing['Temporada'] == season]
    
    # Plot 1: Pass Completions and Interceptions by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data_renamed = season_data.rename(columns={'Cmp': 'Pases completados', 'Int.': 'Pases intentados'})
    season_data_renamed.set_index('Equipo')[['Pases completados', 'Pases intentados']].plot(kind='bar', color=['steelblue', 'coral'], ax=ax)
    ax.set_title(f'Pases Completos y Intercepciones por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Conteo')
    ax.legend()
    ax.tick_params(axis='x', rotation=45)  # Adjust rotation to 45 degrees
    plt.xticks(ha='right')  # Align text to the right
    plt.tight_layout()
    plt.show()

    # Plot 2: Pass Completion Percentage by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    ax.bar(season_data['Equipo'], season_data['% Cmp'], color='lightgreen')
    ax.set_title(f'Porcentaje de Pases Completos por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('% de Pases Completos')
    ax.tick_params(axis='x', rotation=45)
    plt.xticks(ha='right')
    plt.tight_layout()
    plt.show()

    # Plot 3: Total Distance and Progressive Distance by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    ax.plot(season_data['Equipo'], season_data['Dist. tot.'], label='Distancia Total', marker='o', color='purple')
    ax.plot(season_data['Equipo'], season_data['Dist. prg.'], label='Distancia Progresiva', marker='o', color='orange')
    ax.set_title(f'Distancia Total y Distancia Progresiva por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Distancia (metros)')
    ax.legend()
    ax.tick_params(axis='x', rotation=45)
    plt.xticks(ha='right')
    plt.tight_layout()
    plt.show()

    # Plot 4: Expected Assists (xA) and Assisted Goals (xAG) by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    ax.bar(season_data['Equipo'], season_data['xA'], label='Asistencias esperadas', color='lightblue')
    ax.bar(season_data['Equipo'], season_data['xAG'], label='Asistencias esperadas que acaban en gol', color='lightcoral', alpha=0.7, bottom=season_data['xA'])
    ax.set_title(f'Asistencias Esperadas (xA) y Goles Asistidos (xAG) por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Valor')
    ax.legend()
    ax.tick_params(axis='x', rotation=45)
    plt.xticks(ha='right')
    plt.tight_layout()
    plt.show()

# Create an interactive slider widget for selecting the season
season_slider = widgets.SelectionSlider(
    options=seasons,
    description='Temporada:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True
)

# Display the slider and link it to the plotting function
interactive_plot = widgets.interactive(plot_season, season=season_slider)
display(interactive_plot)
```

# Estadisticas de tiro

```python
import matplotlib.pyplot as plt
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# Ensure plots display inline in Jupyter
%matplotlib inline

# Load data
df_shooting = pd.read_csv("C:/UABC/Topicos selectos de la investigacion/MatchesDataSet/data_sets/estadisticas_tiros_2017_2024.txt")
df_shooting['Temporada'] = df_shooting['Temporada'].astype(str)

# Define available seasons
seasons = sorted(df_shooting['Temporada'].unique())

# Define the plotting function for a selected season
def plot_season(season):
    season_data = df_shooting[df_shooting['Temporada'] == season]
    
    # Plot 1: Goals (Gls.) and Shots (Dis) by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data_renamed = season_data.rename(columns={'Gls.': 'Goles marcados', 'Dis': 'Disparos intentados'})
    season_data_renamed.set_index('Equipo')[['Goles marcados', 'Disparos intentados']].plot(kind='bar', color=['steelblue', 'coral'], ax=ax)
    ax.set_title(f'Goles y Disparos por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Conteo')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # Plot 2: Detailed Shooting Breakdown
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data_renamed = season_data.rename(columns={'DaP': 'Disparos a puerta', 'FK': 'Disparos de tiro libre', 'TP': 'Disparos a puerta', 'TPint': 'Disparos a puerta (penales)'})
    season_data_renamed.set_index('Equipo')[['Disparos a puerta', 'Disparos de tiro libre', 'Disparos a puerta', 'Disparos a puerta (penales)']].plot(kind='bar', stacked=True, ax=ax, color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'])
    ax.set_title(f'Detalle de Disparos por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Conteo de Disparos')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # Plot 3: Expected Goals (xG) vs. Non-Penalty Expected Goals (npxG) by Team
    fig, ax = plt.subplots(figsize=(14, 8))
    ax.bar(season_data['Equipo'], season_data['xG'], color='lightblue', label='xG')
    ax.bar(season_data['Equipo'], season_data['npxG'], color='salmon', bottom=season_data['xG'], label='npxG')
    ax.set_title(f'Goles Esperados (xG) y Goles Esperados sin Penales (npxG) por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Valor de Goles Esperados')
    ax.legend()
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()


# Create an interactive slider widget for selecting the season
season_slider = widgets.SelectionSlider(
    options=seasons,
    description='Temporada:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True
)

# Display the slider and link it to the plotting function
interactive_plot = widgets.interactive(plot_season, season=season_slider)
display(interactive_plot)
```

# Estadisticas de porteria

```python
import matplotlib.pyplot as plt
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# Ensure plots display inline in Jupyter
%matplotlib inline

# Load data
df_goalkeeping_stats = pd.read_csv("C:/UABC/Topicos selectos de la investigacion/MatchesDataSet/data_sets/estadisticas_porteros_2017_2024.txt")
df_goalkeeping_stats['Temporada'] = df_goalkeeping_stats['Temporada'].astype(str)

# Define available seasons
seasons = sorted(df_goalkeeping_stats['Temporada'].unique())

# Define the plotting function for a selected season
def plot_season(season):
    season_data = df_goalkeeping_stats[df_goalkeeping_stats['Temporada'] == season]
    
    # Plot 1: Goals Conceded (GC) and Shots on Target Against (DaPC)
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data.set_index('Equipo')[['GC', 'DaPC']].plot(kind='bar', color=['#1f77b4', '#ff7f0e'], ax=ax)
    ax.set_title(f'Goles en Contra (GC) y Disparos a Puerta en Contra (DaPC) por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Conteo')
    ax.text(0, -0.15, 'GC: Goles en contra | DaPC: Disparos a puerta en contra', transform=ax.transAxes, ha='center', fontsize=10, color='gray')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # Plot 2: Saves (Salvadas) and Clean Sheets (PaC)
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data.set_index('Equipo')[['Salvadas', 'PaC']].plot(kind='bar', stacked=True, ax=ax, color=['#2ca02c', '#d62728'])
    ax.set_title(f'Salvadas y Porterías a Cero (PaC) por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Conteo')
    ax.text(0, -0.15, 'Salvadas: Atajadas realizadas por el portero | PaC: Portería a cero', transform=ax.transAxes, ha='center', fontsize=10, color='gray')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # Plot 3: Wins (PG), Draws (PE), and Losses (PP)
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data.set_index('Equipo')[['PG', 'PE', 'PP']].plot(kind='bar', ax=ax, color=['#17becf', '#bcbd22', '#e377c2'])
    ax.set_title(f'Partidos Ganados (PG), Empatados (PE), y Perdidos (PP) por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Número de Partidos')
    ax.text(0, -0.15, 'PG: Partidos ganados | PE: Partidos empatados | PP: Partidos perdidos', transform=ax.transAxes, ha='center', fontsize=10, color='gray')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # Plot 4: Penalty Attempts (TPint) and Penalties Conceded (PD)
    fig, ax = plt.subplots(figsize=(14, 8))
    season_data.set_index('Equipo')[['TPint', 'PD']].plot(kind='bar', color=['#9467bd', '#8c564b'], ax=ax)
    ax.set_title(f'Tiros de Penal Intentados (TPint) y Penales Concedidos (PD) por Equipo ({season})')
    ax.set_xlabel('Equipo')
    ax.set_ylabel('Conteo')
    ax.text(0, -0.15, 'TPint: Tiros de penal intentados | PD: Penales concedidos', transform=ax.transAxes, ha='center', fontsize=10, color='gray')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# Create an interactive slider widget for selecting the season
season_slider = widgets.SelectionSlider(
    options=seasons,
    description='Temporada:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True
)

# Display the slider and link it to the plotting function
interactive_plot = widgets.interactive(plot_season, season=season_slider)
display(interactive_plot)
```
