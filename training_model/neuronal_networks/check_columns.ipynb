```python
import pandas as pd

# Cargar los datasets
betis_vs_atletico = pd.read_csv("C:/UABC/Topicos selectos de la investigacion/MatchesDataSet/current_season_data/j11/betis_atletico.csv")
merged_all_statistics = pd.read_csv("C:/UABC/Topicos selectos de la investigacion/MatchesDataSet/merged_all_statistics.txt", sep=",")

# Obtener los nombres de las columnas
columns_betis_vs_atletico = set(betis_vs_atletico.columns)
columns_merged = set(merged_all_statistics.columns)

# Comparar columnas
faltan_en_betis_vs_atletico = columns_merged - columns_betis_vs_atletico
sobran_en_betis_vs_atletico = columns_betis_vs_atletico - columns_merged

# Resultados
print("Columnas que faltan en betis_vs_atletico:")
print(faltan_en_betis_vs_atletico)

print("\nColumnas adicionales en betis_vs_atletico (no están en merged_all_statistics):")
print(sobran_en_betis_vs_atletico)
```
